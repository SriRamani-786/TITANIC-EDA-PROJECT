```python
from IPython.display import display, Markdown
import pandas as pd

# Prepare dataset
df_clean = df.copy()

# Handle missing values
df_clean['Age'] = df_clean.groupby(
    ['Sex', 'Pclass']
)['Age'].transform(lambda x: x.fillna(x.median()))

df_clean['Embarked'] = df_clean['Embarked'].fillna(
    df_clean['Embarked'].mode()[0]
)

# Create AgeGroup
df_clean['AgeGroup'] = pd.cut(
    df_clean['Age'],
    bins=[0, 12, 18, 35, 60, 100],
    labels=['Child', 'Teenager', 'Young Adult', 'Adult', 'Senior'],
    include_lowest=True
)

# Create FamilySize
df_clean['FamilySize'] = (
    df_clean['SibSp'] + df_clean['Parch'] + 1
)

# Create TravelType
df_clean['TravelType'] = df_clean['FamilySize'].apply(
    lambda x: 'Alone' if x == 1 else 'With Family'
)

# Calculate survival rates
gender_survival = df_clean.groupby('Sex')['Survived'].mean() * 100

class_survival = df_clean.groupby('Pclass')['Survived'].mean() * 100

age_survival = df_clean.groupby(
    'AgeGroup', observed=False
)['Survived'].mean() * 100

family_survival = df_clean.groupby(
    'FamilySize'
)['Survived'].mean() * 100

travel_survival = df_clean.groupby(
    'TravelType'
)['Survived'].mean() * 100

fare_analysis = df_clean.groupby(
    'Survived'
)['Fare'].agg(['mean', 'median'])

# Display Key Findings
key_findings = f"""
## 7. Key Findings

### 7.1 Gender Analysis
- Female Survival Rate: {gender_survival['female']:.2f}%
- Male Survival Rate: {gender_survival['male']:.2f}%

### 7.2 Passenger Class Analysis
- First Class: {class_survival[1]:.2f}%
- Second Class: {class_survival[2]:.2f}%
- Third Class: {class_survival[3]:.2f}%

### 7.3 Age Analysis
- Child: {age_survival['Child']:.2f}%
- Teenager: {age_survival['Teenager']:.2f}%
- Young Adult: {age_survival['Young Adult']:.2f}%
- Adult: {age_survival['Adult']:.2f}%
- Senior: {age_survival['Senior']:.2f}%

### 7.4 Family Analysis
- Travelling Alone: {travel_survival['Alone']:.2f}%
- Travelling With Family: {travel_survival['With Family']:.2f}%

### 7.5 Fare Analysis
- Average Fare (Not Survived): {fare_analysis.loc[0, 'mean']:.2f}
- Average Fare (Survived): {fare_analysis.loc[1, 'mean']:.2f}
- Median Fare (Not Survived): {fare_analysis.loc[0, 'median']:.2f}
- Median Fare (Survived): {fare_analysis.loc[1, 'median']:.2f}

### 7.6 Overall Findings

Gender, passenger class, age, family size, and ticket fare
showed important relationships with Titanic passenger survival.
"""

display(Markdown(key_findings))

# Display numerical outputs
print("GENDER SURVIVAL RATE")
display(gender_survival.round(2))

print("PASSENGER CLASS SURVIVAL RATE")
display(class_survival.round(2))

print("AGE GROUP SURVIVAL RATE")
display(age_survival.round(2))

print("FAMILY SIZE SURVIVAL RATE")
display(family_survival.round(2))

print("TRAVEL TYPE SURVIVAL RATE")
display(travel_survival.round(2))

print("FARE ANALYSIS")
display(fare_analysis.round(2))

print("KEY FINDINGS COMPLETED SUCCESSFULLY!")
```
